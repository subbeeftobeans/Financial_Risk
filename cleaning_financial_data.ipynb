```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_clean = df.dropna() #dropping rows with null values
```

```python
df_clean.to_csv("cleaned_financial_data.csv", index=False) #downloading cleaned CSV 

print(f"Original rows: {len(df)}") # printing original rows of the CSV
print(f"Clean rows: {len(df_clean)}") # printing rows after dropping null values
```
